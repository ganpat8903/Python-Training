### Validate functions

```python
def input_int(prompt):
    while True:
        try:
            n= int(input(prompt))
            if n<0:
                raise Exception("It must be positive!")
            return n
        except ValueError:
            print("".center(100,"-"))
            print("Invalid! Must be a number.")
            print("".center(100,"-"))
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
            
def input_float(prompt):
    while True:
        try:
            n= float(input(prompt))
            if n<0:
                raise Exception("It must be positive!")
            return n
        except ValueError:
            print("".center(100,"-"))
            print("Invalid! Must be a number.")
            print("".center(100,"-"))
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
    
def input_string(prompt):
    while True:
        try:
            s=input(prompt).strip()
            if s=="":
                raise Exception("It must not be empty!")
            return s
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
```

### 2. Given a CSV file with employee details (name, position, salary), create a class to represent an employee.

### 3. Implement a program that simulates a basic bank account using a BankAccount class.


```python
class BankAccount:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance

    def deposit(self):
        amount=input_float("Enter amount to deposit: ")
        self.balance += amount
        print("".center(100,"~"))
        print(f"Deposited Rs{amount:.2f}. New balance: Rs{self.balance:.2f}")
        print("".center(100,"~"))

    def withdraw(self):
        amount=input_float("Enter amount to withdraw: ")
        if amount > self.balance:
            print("".center(100,"-"))
            print("Insufficient balance.")
            print("".center(100,"-"))
        else:
            self.balance -= amount
            print("".center(100,"~"))
            print(f"Withdraw Rs{amount:.2f}. New balance: Rs{self.balance:.2f}")
            print("".center(100,"~"))

    def get_balance(self):
        return f"Current Balance: Rs{self.balance:.2f}"

name=input_string("Enter your name: ")   

account = BankAccount(name)
print(account.get_balance())                

while True:
    print("Enter 1 to deposit.")
    print("Enter 2 to withdraw.")
    print("Enter 3 to display current balance.")
    print("Enter 4 to 'exit'.")
    try:
        x=int(input("Choose from above number: "))

        match x: 
            case 1:account.deposit()
                
            case 2:account.withdraw()
                
            case 3:
                print("".center(100,"~"))
                print(account.get_balance())
                print("".center(100,"~"))
                
            case 4:
                print("Thank You!")
                break
                
            case _:
                print("".center(100,"-"))
                print("Enter a valid number!")
                print("".center(100,"-"))
                
    except ValueError:
        print("".center(100,"-"))
        print("Invalid! Choose again..")
        print("".center(100,"-"))
    
```

### 4. Write a Python program that uses a Rectangle class to calculate the area and perimeter of a rectangle.


```python
class Rectangle:
    
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
        
    def calculate_area(self):
        area=self.length * self.breadth
        return f"Area of Rectangle: {area:.2f}"
        
    def calculate_perimeter(self):
        perimeter=2*(self.length + self.breadth)
        return f"Perimeter of Rectangle: {perimeter:.2f}"
    
length=input_float("Enter length: ")
breadth=input_float("Enter breadth: ")
        
rectangle=Rectangle(length,breadth)
print(rectangle.calculate_area())
print(rectangle.calculate_perimeter())
```

### 6. Given a JSON file with customer data, create a Customer class to store and manipulate the data.

```python
import json
class Customer:
       
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def display(self):
        print("".center(100,"~"))
        print(f"Name: {self.name} \nAge: {self.age}")
        print("".center(100,"~"))
    
    def change_name(self,new_name):
        self.name=new_name
        
    def change_age(self,new_age):
        self.age=new_age
    
customers=[]
try:
    with open("customer-6.json","r") as file:
        data=json.load(file)
        print(data)
        try:
            for item in data:
                customer=Customer(item["name"],item["age"])
                customers.append(customer)
        except ValueError as error:
            print("Invalid data",error)
        
except FileNotFoundError:
    print("File not found!")
    
new_customers=[]
for customer in customers:
    customer.display()
    name=input_string("Enter new name for above customer:")
    customer.change_name(name)
    age=input_int("Enter new age for above customer:")
    customer.change_age(age)
    new_customer={}
    new_customer["name"]=name
    new_customer['age']=age
    new_customers.append(new_customer)
print(new_customers)
try:
    with open("customer-6.json","w") as file:
        file.write(json.dumps(new_customers,indent=5))
        
except FileNotFoundError:
    print("File not found!")
```

### 8. Implement a program that uses a Circle class to calculate the area and circumference of multiple circles3


```python
import math
class Circle:
    
    def __init__(self,radius):
        self.radius=radius
        
    def calculate_area(self):
        area=math.pi * self.radius * self.radius
        return f"{area:.2f}"
    
    def calculate_circumference(self):
        circumference=2 * math.pi * self.radius
        return f"{circumference:.2f}"

number=input_int("Enter number of circles you want: ")
for index in range(number):
    
    radius=input_float(f"Enter radius for circle {index+1}: ")
        
    circle=Circle(radius)
    print(circle.calculate_area())
    print(circle.calculate_circumference())
```

### 11. Create a base class Shape with methods to calculate area and perimeter and derive classes Circle and Square.


```python
import math
class Shape:
    
    def calculate_area_of_circle(self):
        area = math.pi * self.radius * self.radius
        return f"Area of Circle: {area:.2f}"
    
    def calculate_area_of_square(self):
        area = self.side * self.side
        return f"Area of Square: {area:.2f}"
    
    def calculate_perimeter_of_circle(self):
        perimeter= 2 * math.pi * self.radius
        return f"Perimeter of Circle: {perimeter:.2f}"
    
    def calculate_perimeter_of_square(self):
        perimeter= 4 * self.side
        return f"Perimeter of Square: {perimeter:.2f}"
    
class Circle(Shape):
    
    def __init__(self,radius):
        self.radius=radius
    
class Square(Shape):
    
    def __init__(self,side):
        self.side=side
        
radius=input_float(f"Enter radius for circle: ")
            
circle=Circle(radius)
print(circle.calculate_area_of_circle())
print(circle.calculate_perimeter_of_circle())

side=input_float(f"Enter side for Square: ")
            
square=Square(side)
print(square.calculate_area_of_square())
print(square.calculate_perimeter_of_square())
```

### 12. Implement a class hierarchy to represent different types of employees (Manager, Engineer) with their attributes.


```python
class Employee:
    
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
        
    def display(self):
        print(f"Name: {self.name} \nSalary: {self.salary}")
        
class Manager(Employee):
    
    def __init__(self,name,salary,type_of_manager):
        super().__init__(name,salary)
        self.type_of_manager=type_of_manager
        
    def display(self):
        print("Manager Details:")
        super().display()
        print(f"type of manager: {self.type_of_manager}")
        
class Engineer(Employee):
    
    def __init__(self,name,salary,type_of_engineer):
        super().__init__(name,salary)
        self.type_of_engineer=type_of_engineer
        
    def display(self):
        print("Engineer Details:")
        super().display()
        print(f"Type of Engineer: {self.type_of_engineer}")
        
manager_name=input_string("Enter Manager Name: ")
manager_salary=input_float("Enter Manager Salary: ")
type_of_manager=input_string("Enter type of Manager: ")

manager=Manager(manager_name,manager_salary,type_of_manager)
manager.display()

engineer_name=input_string("Enter Engineer Name: ")
engineer_salary=input_float("Enter Engineer Salary: ")
type_of_engineer=input_string("Enter type of Engineer: ")

engineer=Engineer(engineer_name,engineer_salary,type_of_engineer)
engineer.display()
```
