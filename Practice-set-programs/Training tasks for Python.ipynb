```python
class Stack:
    
    def __init__(self):
        self.stack=[]
    
    def push(self,element):
        self.stack+=[element]
        print("".center(100,"~"))
        print(f"{element} is added.")
        print("".center(100,"~"))
    
    def pop(self):
        try:
            print("".center(100,"~"))
            print(f"{self.stack[-1]} is deleted.")
            print("".center(100,"~"))
            self.stack[:]=self.stack[:-1]
            
        except Exception as error:
            print("".center(100,"-"))
            print("Stack is empty!",error)
            print("".center(100,"-"))
        
    def peek(self):
        try:
            print("".center(100,"~"))
            print(f"peek: {self.stack[-1]}")
            print("".center(100,"~"))
            
        except Exception as error:
            print("".center(100,"-"))
            print("Stack is empty!",error)
            print("".center(100,"-"))
    
    def display(self):
        print("".center(100,"~"))
        print("Stack=",self.stack)
        print("".center(100,"~"))
    
stack_object=Stack()

while True:
    print("Enter 1 to push.")
    print("Enter 2 to pop.")
    print("Enter 3 to peek.")
    print("Enter 4 to display.")
    print("Enter 5 to 'exit'.")
    
    choice=input_int("Enter your choice: ")
    
    match choice:
        case 1:
            n=input_int("Enter a number: ")
            stack_object.push(n)           
            
        case 2:stack_object.pop()
            
        case 3:stack_object.peek()
        
        case 4:stack_object.display()
            
        case 5:
            print("Thank You!")
            break
            
        case _:
            print("".center(100,"-"))
            print("Invalid Choice!")
            print("".center(100,"-"))
```

```python
class Queue:
    
    def __init__(self):
        self.queue=[]
        
    def enqueue(self,element):
        self.queue+=[element]
        print("".center(100,"~"))
        print(f"{element} is added.")
        print("".center(100,"~"))
        
    def dequeue(self):
        try:
            print("".center(100,"~"))
            print(f"{self.queue[0]} is deleted.")
            print("".center(100,"~"))
            self.queue[:]=self.queue[1:]
                
        except Exception as error:
            print("".center(100,"-"))
            print("Stack is empty!",error)
            print("".center(100,"-"))
            
    def display(self):
        print("".center(100,"~"))
        print("Queue=",self.queue)
        print("".center(100,"~"))
    
queue_object=Queue()

while True:
    print("Enter 1 to enqueue.")
    print("Enter 2 to dequeue.")
    print("Enter 3 to display.")
    print("Enter 4 to 'exit'.")
    
    choice=input_int("Enter your choice: ")
    
    match choice:
        case 1:
            n=input_int("Enter a number: ")
            queue_object.enqueue(n)
            
        case 2:queue_object.dequeue()
            
        case 3:queue_object.display()
           
        case 4:
            print("Thank You!")
            break
            
        case _:
            print("".center(100,"-"))
            print("Invalid Choice!")
            print("".center(100,"-"))
```

### Selection Sort

```python
list_numbers=input("Enter numbers: ").split()
list_numbers=list(map(int,list_numbers))
print(list_numbers)

for index1 in range(len(list_numbers)):
    
    minimum_number_index=index1
    for index2 in range(index1,len(list_numbers)):
        
        if list_numbers[index2]<list_numbers[minimum_number_index]:
            minimum_number_index=index2
    
    list_numbers[index1],list_numbers[minimum_number_index]=list_numbers[minimum_number_index],list_numbers[index1]

print(list_numbers)
```

### Bubble Sort

```python
list_numbers=input("Enter numbers: ").split()
list_numbers=list(map(int,list_numbers))
print(list_numbers)

for index1 in range(len(list_numbers)):
    
    for index2 in range(len(list_numbers)-index1-1):
        
        if list_numbers[index2]>list_numbers[index2+1]:
            list_numbers[index2],list_numbers[index2+1]=list_numbers[index2+1],list_numbers[index2]
            
print(list_numbers)
```

### Insertion Sort

```python
list_numbers=input("Enter numbers: ").split()
list_numbers=list(map(int,list_numbers))
print(list_numbers)

for index in range(1, len(list_numbers)):
    key = list_numbers[index]
    j = index - 1

    while j >= 0 and list_numbers[j] > key:
        list_numbers[j + 1] = list_numbers[j]
        j -= 1
    list_numbers[j + 1] = key

print(list_numbers)
```

### Quick Sort

```python
def quick_sort(list_numbers):
    
    if len(list_numbers)==0:
        return []
    
    pivot = list_numbers[0]
    less = []
    greater = []

    for index in range(1,len(list_numbers)):
        
        if list_numbers[index] <= pivot:
            less.append(list_numbers[index])
        else:
            greater.append(list_numbers[index])

    return quick_sort(less) + [pivot] + quick_sort(greater)

list_numbers=input("Enter numbers: ").split()
list_numbers=list(map(int,list_numbers))
print(list_numbers)

print(quick_sort(list_numbers))
```

### Binary Search

```python
list_numbers=input("Enter a sorted array : ").split()
list_numbers=list(map(int,list_numbers))
print(list_numbers)

target=input_int("Enter a number to find: ")

low = 0
high = len(list_numbers) - 1

while low <= high:
    
    mid = (low + high) // 2

    if list_numbers[mid] == target:
        print(f"Element found at index {mid}")
        break
    
    elif list_numbers[mid] < target:
        low = mid + 1
        
    else:
        high = mid - 1

else:
     print("Element not found")
```

### W.A.P. that replicates a complete banking system. User should be able to perform below given tasks.

- __Open account__
- __login using account number and password__
- __logout__
- __add/withdraw money__
- __close account__
- __transfer amount__
- __keep track of transactions (all)__
- __maintain total amount of money that bank has__
- __take loan from bank at 7% interest rate for a time periode and pay EMIs__
- __Do this without an external database__
- __Use a JSON file as database__
- __Use postgresql as database__


```python
class Bank:
    
    def __init__(self):
        self.accounts = {} 
        self.logged_in_user = None
        self.bank_total_balance = 0
        self.loan_interest_rate = 0.07
        self.account_counter = 1001 

    def open_account(self):
        try:
            name = input("Enter your name: ")
            password = input("Set your password: ")
            initial_deposit = float(input("Enter initial deposit amount: "))
            account_number = self.account_counter
            
            self.accounts[account_number] = {
                "name": name,
                "password": password,
                "balance": initial_deposit,
                "transactions": [f"Account opened with {initial_deposit}"],
                "loan": None
            }
            
            self.bank_total_balance += initial_deposit
            self.account_counter += 1
            
            print(f"Account created successfully. Your account number is: {account_number}")
            
        except Exception as e:
            print("Error creating account:", e)

    def login(self):
        try:
            account_number = int(input("Enter your account number: "))
            password = input("Enter your password: ")
            
            if account_number in self.accounts and self.accounts[account_number]["password"] == password:
                self.logged_in_user = account_number
                print("Login successful.")
            else:
                print("Invalid data!.")
                
        except Exception as e:
            print("Login error:", e)

    def logout(self):
        
        self.logged_in_user = None
        print("Logged out successfully.")

    def add_money(self):
        try:
            if self.logged_in_user:
                amount = float(input("Enter amount to deposit: "))
                self.accounts[self.logged_in_user]["balance"] += amount
                self.accounts[self.logged_in_user]["transactions"].append(f"Deposited {amount}")
                self.bank_total_balance += amount
                print("Amount added successfully.")
            else:
                print("Please login first.")
                
        except Exception as e:
            print("Error adding money:", e)

    def withdraw_money(self):
        try:
            if self.logged_in_user:
                amount = float(input("Enter amount to withdraw: "))
                if self.accounts[self.logged_in_user]["balance"] >= amount:
                    self.accounts[self.logged_in_user]["balance"] -= amount
                    self.accounts[self.logged_in_user]["transactions"].append(f"Withdraw {amount}")
                    self.bank_total_balance -= amount
                    print("Amount withdrawn successfully.")
                else:
                    print("Insufficient balance.")
            else:
                print("Please login first.")
        except Exception as e:
            print("Error withdrawing money:", e)

    def transfer_amount(self):
        try:
            if self.logged_in_user:
                target = int(input("Enter target account number: "))
                amount = float(input("Enter amount to transfer: "))
                
                if target in self.accounts:
                    
                    if self.accounts[self.logged_in_user]["balance"] >= amount:
                        self.accounts[self.logged_in_user]["balance"] -= amount
                        self.accounts[target]["balance"] += amount
                        self.accounts[self.logged_in_user]["transactions"].append(f"Transferred {amount} to {target}")
                        self.accounts[target]["transactions"].append(f"Received {amount} from {self.logged_in_user}")
                        print("Transfer successful.")
                    else:
                        print("Insufficient balance.")
                        
                else:
                    print("Target account not found.")
                    
            else:
                print("Please login first.")
        except Exception as e:
            print("Error during transfer:", e)

    def close_account(self):
        if self.logged_in_user:
            data = self.accounts.pop(self.logged_in_user)
            self.bank_total_balance -= data["balance"]
            print(f"Account closed. Returned balance: {data['balance']}")
            self.logged_in_user = None
        else:
            print("Please login first.")

    def view_transactions(self):
        if self.logged_in_user:
            print("Transaction history:")
            for transaction in self.accounts[self.logged_in_user]["transactions"]:
                print("-", transaction)
        else:
            print("Please login first.")

    def view_balance(self):
        if self.logged_in_user:
            print("Your balance is:", self.accounts[self.logged_in_user]["balance"])
        else:
            print("Please login first.")

    def view_bank_balance(self):
        print("Bank total money:", self.bank_total_balance)

    def take_loan(self):
        try:
            if self.logged_in_user:
                
                if self.accounts[self.logged_in_user]["loan"]:
                    print("You already have an active loan.")
                    return
                
                amount = float(input("Enter loan amount: "))
                years = int(input("Enter loan duration in years: "))
                
                interest = amount * self.loan_interest_rate * years
                total = amount + interest
                emi = total / (years * 12)
                
                self.accounts[self.logged_in_user]["loan"] = {
                    "amount": amount,
                    "years": years,
                    "emi": round(emi, 2),
                    "remaining_months": years * 12,
                    "remaining_amount": round(total, 2)
                }
                
                self.accounts[self.logged_in_user]["balance"] += amount
                self.bank_total_balance -= amount
                self.accounts[self.logged_in_user]["transactions"].append(f"Took loan of {amount}, EMI: {round(emi, 2)}")
                print(f"Loan granted. EMI per month: {round(emi, 2)} for {years * 12} months")
                
            else:
                print("Please login first.")
                
        except Exception as e:
            print("Error taking loan:", e)

    def pay_emi(self):
        
        if self.logged_in_user:
            loan = self.accounts[self.logged_in_user]["loan"]
            
            if not loan:
                print("You have no active loan.")
                return
            
            if self.accounts[self.logged_in_user]["balance"] >= loan["emi"]:
                
                self.accounts[self.logged_in_user]["balance"] -= loan["emi"]
                self.bank_total_balance += loan["emi"]
                loan["remaining_amount"] -= loan["emi"]
                loan["remaining_months"] -= 1
                self.accounts[self.logged_in_user]["transactions"].append(f"Paid EMI of {loan['emi']}")
                print(f"EMI paid. Remaining amount: {round(loan['remaining_amount'], 2)}")
                
                if loan["remaining_months"] <= 0:
                    print("Loan fully repaid.")
                    self.accounts[self.logged_in_user]["loan"] = None
            else:
                print("Not enough balance to pay EMI.")
        else:
            print("Please login first.")


bank = Bank()
print(bank.accounts)
while True:
    print("".center(100,"-"))
    print("1. Open Account")
    print("2. Login")
    print("3. Logout")
    print("4. Add Money")
    print("5. Withdraw Money")
    print("6. Transfer Money")
    print("7. View Transactions")
    print("8. View Balance")
    print("9. Close Account")
    print("10. Bank Total Money")
    print("11. Take Loan")
    print("12. Pay EMI")
    print("13. Exit")
    print("".center(100,"-"))

    choice = input("Enter your choice: ")

    match choice:
        case '1':
            bank.open_account()
        case '2':
            bank.login()
        case '3':
            bank.logout()
        case '4':
            bank.add_money()
        case '5':
            bank.withdraw_money()
        case '6':
            bank.transfer_amount()
        case '7':
            bank.view_transactions()
        case '8':
            bank.view_balance()
        case '9':
            bank.close_account()
        case '10':
            bank.view_bank_balance()
        case '11':
            bank.take_loan()
        case '12':
            bank.pay_emi()
        case '13':
            print("Thanks for using our banking system!")
            break
        case _:
            print("Invalid choice! Try again.")
```
