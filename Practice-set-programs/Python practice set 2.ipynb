### 1. Given two lists of numbers, concatenate them into a single list

### 2. Write a program that finds the largest and smallest elements in a list


```python
numbers=input("Enter list of numbers: ").split()
numbers=list(map(int,numbers))
minimum_number=numbers[0]
maximum_number=numbers[0]
for index in range(1,len(numbers)):
    if numbers[index]<minimum_number:
        minimum_number=numbers[index]
    if numbers[index]>maximum_number:
        maximum_number=numbers[index]
print(f"Smallest number: {minimum_number} \nLargest number: {maximum_number}")
```

### 3. Implement a function that takes a list of numbers and returns a new list with the squared values


```python
def square(numbers):
    new_numbers=[]
    for number in numbers:
        new_numbers.append(number*number)
    return new_numbers
numbers=input("Enter list of numbers: ").split()
numbers=list(map(int,numbers))
print(f"Before square: {numbers}")
print(f"After square: {square(numbers)}")
```

### 5. Given a list of words, find the word with the maximum length and its length

### 8. Create a function that takes a list of strings and returns the list sorted by the length of the strings

```python
def sorted_list(strings):
    for index1 in range(len(strings)):
        for index2 in range(index1+1,len(strings)):
            if len(strings[index1])>len(strings[index2]):
                strings[index1],strings[index2]=strings[index2],strings[index1]
    return strings
strings=input("Enter list of strings: ").split()

print(f"Original strings: {strings}")
print(f"sorted strings: {sorted_list(strings)}")
```

### 10. Implement a function that takes two lists and returns their union (all unique elements from both lists).


```python
def union_of_list(lists1,lists2):
    lists=[]
    for element in lists1:
        if element not in lists:
            lists.append(element)
    for element in lists2:
        if element not in lists:
            lists.append(element)
    return lists
lists1=input("Enter number of lists 1: ").split()
lists2=input("Enter number of lists 2: ").split()
print(union_of_list(lists1,lists2))
```

### 18. Create a function that takes a list of dictionaries and sorts them based on a specified key


```python
def sort_salary(dictionary_list):
    for index in range(len(dictionary_list)-1):
        for count in range(index+1,len(dictionary_list)):
            if dictionary_list[index]['salary']>dictionary_list[count]['salary']:
                    dictionary_list[index],dictionary_list[count]=dictionary_list[count],dictionary_list[index]
    return dictionary_list

dictionary_list=[]
number=int(input("Enter number of dictionary you want in list: "))
for index in range(number):
    print(f"Enter dictionary {index+1}:".center(100,"-"))
    dictionary={}
    name=input("Enter name: ").strip()
    salary=int(input("Enter salary: "))
    dictionary['name']=name
    dictionary['salary']=salary
    dictionary_list.append(dictionary)
print(f"Before sorted salary: {dictionary_list}")
print(f"After Sorted salary: {sort_salary(dictionary_list)}")
```

```python
def sort_salary(dictionary_list,key):
    return sorted(dictionary_list,key=lambda x:x.get(key,0))

dictionary_list=[]
number=int(input("Enter number of dictionary you want in list: "))
for index in range(number):
    print(f"Enter dictionary {index+1}:".center(100,"-"))
    dictionary={}
    name=input("Enter name: ").strip()
    salary=int(input("Enter salary: "))
    dictionary['name']=name
    dictionary['salary']=salary
    dictionary_list.append(dictionary)
print(f"Before sorted salary: {dictionary_list}")
print(f"After Sorted salary: {sort_salary(dictionary_list,'salary')}")
```

### 19. Write a program that finds the average value of all the elements in a list of dictionaries

### 20. Implement a function that takes a list of strings and returns a set of unique characters in all strings.


```python
def unique_characters(strings):
    for element in strings:
        unique_letters = [char for char in element if element.count(char) == 1]
        print(",".join(unique_letters)) 
        
strings = input("Enter list of strings: ").split()
unique_characters(strings)
```
