### 6. Calculate the compound interest for a given principal amount, interest rate, and time period

```python
# CI = P(1 + r/100)n – P
amount=float(input("Enter amount: "))
rate=float(input("Enter interest rate: "))
time=int(input("Enter time period: "))
compound_interest= amount*(1+rate/100)**time - amount
print(f"compound interest: {compound_interest:.2f}")
```

### 16. Write a Python program to check if a given string is a pangram (contains all letters of the alphabet)


```python
# a sentence is a pangram if it contains each of the 26 letters in the English alphabet.it is not case sensitive
letters="abcdefghijklmnopqrstuvwxyz"
string=input("Enter a string: ")
count=0
string_set=set(string)
string_set.discard(" ")
if len(string_set)>=26:
    for letter in letters:
        if letter in string.lower():
            count+=1
    if count==26:
        print("string is a pangram")
    else:
        print("string is not a pangram")    
else:
    print("string is not a pangram")  
    
    
# Optimize
```

### 20. Write a program to check if a number is prime.


```python
number=int(input("Enter a number: "))
if number>1:
    for num in range(2,number//2):
        if number%num==0:
            print(f"{number} is not prime")
            break
    else:
        print(f"{number} is prime")
else:
    print(f"{number} is not prime")
```

### 21. Create a program that takes a year as input and checks if it is a leap year or not


```python
year=int(input("Enter year: "))
if year%4==0 and (year%100!=0 or year%400==0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")    
```

### 27. Write a program that calculates the factorial of a given number

```python
def find_factorial(num):
    if num<=1:
        return 1
    else:
        return num*find_factorial(num-1)
num=int(input("Enter a number: "))
print(f"Factorial of {num} is {find_factorial(num)}")
```

### 30. Calculate the sum of digits of a given number.


```python
number=int(input("Enter a number: "))
num=number
sum=0
while num>0:
    remainder=num%10
    sum=sum+remainder
    num//=10
print(f"sum of digits of number {number} is {sum}")
```
