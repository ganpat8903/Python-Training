### Validate functions

```python
def input_int(prompt):
    while True:
        try:
            n= int(input(prompt))
            if n<0:
                raise Exception("It must be positive!")
            return n
        except ValueError:
            print("".center(100,"-"))
            print("Invalid! Must be a number.")
            print("".center(100,"-"))
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
            
def input_float(prompt):
    while True:
        try:
            n= float(input(prompt))
            if n<0:
                raise Exception("It must be positive!")
            return n
        except ValueError:
            print("".center(100,"-"))
            print("Invalid! Must be a number.")
            print("".center(100,"-"))
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
    
def input_string(prompt):
    while True:
        try:
            s=input(prompt).strip()
            if not s:
                raise Exception("It must not be empty!")
            return s
        except Exception as error:
            print("".center(100,"-"))
            print(error)
            print("".center(100,"-"))
```

### 1. Write a Python program to copy the contents of one text file into another

```python
try:
    
    with open("file1.txt") as file_1:
        data=file_1.read().strip()
        print(data)
        
    with open("file2.txt","w") as file_2:
        file_2.write(data)
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 2. Given a CSV file with student names and scores, find the student with the highest score


```python
try:
    
    with open("student-2.csv") as file:
        lines=file.readlines()[1:]
        print(lines)
        
        high_score=0
        student_name=None
        for line in lines:
            name,score=line.strip().split(",")
            score=int(score)
            if score>high_score:
                high_score=score
                student_name=name
        print(f"Student name: {student_name} \nwith highest score: {high_score}")
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 3. Implement a program that reads a text file and counts the number of words and lines in it


```python
try:
    
    with open("file1.txt") as file_1:
        data=file_1.read().strip()
        print(data)
        
        words=data.split()
        lines = [line for line in data.split("\n") if line.strip() != ""]
       
        print(f"total words: {len(words)}")
        print(f"total lines: {len(lines)}")
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 4. Create a function that takes a list of sentences and writes them to a new text file, each on a new line


```python
def write_file(list_sentences):
    
    try:
    
        with open("newfile-4.txt","a") as file:
            for sentence in list_sentences:
                file.write(sentence+"\n")
        
    except FileNotFoundError:
        print("File not found!")
    
    except Exception as error:
        print("Invalid!",error)

list_sentences=[]
number=input_int("Enter number of sentences you want: ")
for index in range(number):
    sentence=input_string(f"Enter sentence: {index+1}")
    list_sentences.append(sentence)

write_file(list_sentences)
```

### 5. Given a CSV file with employee details (name, age, salary), calculate the average salary of all employees


```python
try:
    
    with open("employee-5.csv") as file:
        lines=file.readlines()[1:]
        print(lines)
        
        sum_salary=0
        for line in lines:
            name,age,salary=line.strip().split(",")
            salary=float(salary)
            sum_salary+=salary
        
        average_salary=sum_salary/len(lines)
        print(f"Average Salary of all employees: {average_salary:.2f}")
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 6. Write a program that reads a CSV file and finds the total sales revenue for a specific product


```python
try:
    
    with open("product-6.csv") as file:
        lines=file.readlines()[1:]
        print(lines)
        
        products={}
        for line in lines:
            product_name,sales=line.strip().split(",")
            sales=int(sales)
            
            if product_name not in products:
                products[product_name]=0
            products[product_name]+=sales
            
        print(products)
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 7. Given a text file with a list of numbers, write a function that finds the sum of all numbers in the file


```python
try:
    
    with open("numbers-7.txt") as file:
        data=file.read().split("\n")
        print(data)
        
        sum_numbers=0
        for item in data:
            number=int(item)
            sum_numbers+=number
            
        print(f"Sum of all numbers: {sum_numbers}")
        
except FileNotFoundError:
    print("File not found!")
    
except Exception as error:
    print("Invalid!",error)
```

### 9. Write a function that reads a JSON file and extracts specific information from it


```python
import json
def read_json_file(file_name):
    
    try:
    
        with open(file_name) as file:
            data=json.load(file)
            
            for item in data:
                print(f"name: {item['name']}, price: {item['price']}, quantity: {item['quantity']}")
        
    except FileNotFoundError:
        print("File not found!")
    
    except Exception as error:
        print("Invalid!",error)
    
read_json_file("product-15.json")
```
