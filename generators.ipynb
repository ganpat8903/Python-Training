```python
def message():
    yield "hii"
    yield "byy"
    
for i in message():
    print(i)
```

```python
def even(nums):
    for n in nums:
        if n%2==0:
            yield n
    
nums=10,21,11,12,13,14
for n in even(nums):
    print(n)
```
