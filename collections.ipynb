```python
import collections 

dict1 = { 'a' : 1, 'b' : 2 } 
dict2 = { 'b' : 3, 'c' : 4 } 
dict3 = { 'c' : 5, 'd' : 6 } 

chain = collections.ChainMap(dict1,dict2,dict3) 
print(chain)

# printing chainMap using maps 
print ("All the ChainMap contents are : ") 
print (chain.maps) 
  
# printing keys using keys() 
print ("All keys of ChainMap are : ") 
print (list(chain.keys())) 
  
# printing keys using keys() 
print ("All values of ChainMap are : ") 
print (list(chain.values())) 
```

```python
class DeepChainMap(collections.ChainMap):
    'Variant of ChainMap that allows direct updates to inner scopes'

    def __setitem__(self, key, value):
        for mapping in self.maps:
            if key in mapping:
                mapping[key] = value
                return
        self.maps[0][key] = value

    def __delitem__(self, key):
        for mapping in self.maps:
            if key in mapping:
                del mapping[key]
                return
        raise KeyError(key)
        

d = DeepChainMap({'zebra': 'black'}, {'elephant': 'blue'}, {'lion': 'yellow'})
print(d)
d['lion'] = 'orange'  # Update at deeper lever

d['snake'] = 'red'    # Adds a new element

del d['elephant']     # Delete at deeper level

print(d)

# You can always change the logic in overriden methods and get desired behaviour
```

```python
c = collections.Counter(a=3, b=1,c=5)
d = collections.Counter(a=1, b=2,d=3)

print(f"c + d: {c + d}")                       # add two counters together:  c[x] + d[x]


print(f"c - d: {c - d}")                       # subtract (keeping only positive counts)


print(f"c & d: {c & d}")                       # intersection:  min(c[x], d[x]) 


print(f"c | d: {c | d}")                       # union:  max(c[x], d[x])
```

```python
from collections import deque

# d = deque('ghi',5) # maxlen
d = deque('ghi')                 # make a new deque with three items
print(d)
print("Iterating over deque: ")
for element in d:                   # iterate over the deque's elements
    print(element.upper(), end=' ')



d.append('j')                    # add a new entry to the right side
print(f"\nd.append('j'): {d}")
d.appendleft('f')                # add a new entry to the left side
print(f"d.appendleft('f'): {d}")                                # show the representation of the deque


d.pop()                        # return and remove the rightmost item
print(f"\nd.pop(): {d}")
d.popleft()                      # return and remove the leftmost item
print(f"d.popleft(): {list(d)}\n")  # list the contents of the deque
                          

print(f"d[0]: {d[0]}")                             # peek at leftmost item

print(f"d[-1]: {d[-1]}\n")                            # peek at rightmost item


print(f"list(reversed(d)): {list(reversed(d))}\n")                # list the contents of a deque in reverse

print(f'"h" in d: {"h" in d}\n')                                # search the deque

print(f"d.extend('jkl'): {d.extend('jkl')}")                  # add multiple elements at once
print(d)

print(f"\nd.rotate(1): {d.rotate(1)}")                      # right rotation
print(d)

print(f"d.rotate(-1): {d.rotate(-1)}")                     # left rotation
print(d)


print(f"\ndeque(reversed(d)): {deque(reversed(d))}\n")               # make a new deque in reverse order

print("count(g):",d.count("g")) # return 0 if no element present
print("index(g):",d.index('g')) #  raises ValueError if not found.

print(f"\ninsert(2,r): {d.insert(2,'r')}")
print(d)

print(f"\nremove(r): {d.remove('r')}") # Remove the first occurrence of value. If not found, raises a ValueError.
print(d)

print(f"\nd.clear(): {d.clear()}")                        # empty the deque
print(d)
```

```python
s = [('yellow', 1), ('blue', 2), ('yellow', 3), ('blue', 4), ('red', 1)]
d = collections.defaultdict(list)
print(d)

for k, v in s:
    d[k].append(v)
    
print(d)
print(d["abc"])
print(d)
print(sorted(d.items()))
```

```python
s = 'mississippi'
d = collections.defaultdict(int)
print(d)

for k in s:
    d[k] += 1

print(d)
print(d["asdf"])
print(sorted(d.items()))
```

```python
def constant_factory(value):
    return lambda: value

d = collections.defaultdict(constant_factory("10"), [('a', 1), ('b', 2)])
print(d)

print(d['a'])
print(d['b'])
print(d['c'])
print(d['d'])
```

```python
cls_obj = collections.namedtuple('Student',['name','age','DOB'])  # Student is the user defined type

print(cls_obj)

S = cls_obj('Nandini','19','2541997')
print(S)

# Access using index 
print ("The Student age using index is : ",end ="") 
print (S[1]) 
  
# Access using name  
print ("The Student name using keyname is : ",end ="") 
print (S.name) 
  
# Access using getattr() 
print ("The Student DOB using getattr() is : ",end ="") 
print (getattr(S,'DOB')) 
```
